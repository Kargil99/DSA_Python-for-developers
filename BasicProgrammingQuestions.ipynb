1.**Count** **Frequency** **of** **each** **element**


```python
def getfreq(arr):
  store = {}
  for num in arr:
    if num in store:
      store[num] += 1
    else:
      store[num] = 1

  for key, val in store.items():
    print(key, val)
```

```python
arr1 = [1,1,2,2,2,2,4,4,4]
print(getfreq(arr1))
```

2. Array reverse

```python
def arrayrev(arr):
  # we can use list slicing method :arr1 = arr[::-1]
  # we can use reverse method as: arr.reverse()

  start, end = 0, len(arr)-1
  while start <= end:
    arr[start], arr[end] = arr[end], arr[start]
    start += 1
    end -= 1
  return arr
```

```python
arr10 = [3,6,7,3,5,6]
print(arrayrev(arr10))
```

**3. Rotate Array by K element:**

```python
def rotatearr(arr,k):
  n = len(arr)
  ans = []
  for i in range(n):
    ans.append(arr[(i+k)%n])

  return ans
```

```python
arr1 = [1,2,3,4,5,6,7]
k =2
print(rotatearr(arr1,k))
```

**4. Find Median of a given array**

```python
def findmedian(arr):
  if len(arr) % 2 == 0:
    ind1 = len(arr)//2
    ind2 = ind1 - 1
    median = (arr[ind1] + arr[ind2])/2
  else:
    median = arr[len(arr)//2]
  return median
```

```python
print(findmedian([1,2,3,4,5]))
print(findmedian([1,2,3,4,5,6,7,8]))
```

**5. Remove duplicate from sorted array**

```python
def removedup(arr):
  store = []
  for num in arr:
    if num not in store:
      store.append(num)
  return store
```

```python
arri = [1,1,2,2,2,2,4,4,4]
print(removedup(arri))
```

**6. Find all non repeating elements in an array:**

```python
def getnonrep(arr):
  store = {}
  for num in arr:
    if num in store:
      store[num] += 1
    else:
      store[num] = 1


  ans = []
  for key, val in store.items():
    if val == 1:
      ans.append(key)
  return ans
```

```python
arr3 = [2,2,2,3,3,4]
print(getnonrep(arr3))
```

**7. Maximum product subarray**

```python
def maxproduct(arr):
  n = len(arr)
  pre, suf = 1, 1
  ans = float('-inf')
  for i in range(n):
    if pre == 0:
      pre = 1
    if suf == 0:
      suf = 1
    pre *= arr[i]
    suf *= arr[n-i-1]
    ans = max(ans, pre, suf)
```

```python
k = maxproduct([1,2,3,4,5])
print(k)
```

**8. Bubble Sort**

```python
def bubblesort(arr):
  for i in range(len(arr)-1):
    for j in range(len(arr)-1-i):
      if arr[j] > arr[j+1]:
        arr[j], arr[j+1] = arr[j+1], arr[j]
  return arr
```

```python
arr4 = [738,3,234,23,2]
print(bubblesort(arr4))
```

**9. Insertion Sort**

```python
def insertionsort(arr):
  n = len(arr)
  for i in range(1,n): #traverse from second element to end of the array
    key = arr[i]   #current element to be compared
    j = i-1
    while j >= 0 and key < arr[j]:
      arr[j+1] = arr[j]
      j -= 1
    arr[j+1] = key
  return arr
```

```python
arr5 = [1,45,67,32,24,89]
print(insertionsort(arr5))
```

**10. Merge sort**

```python
def mergesort(arr):
  if len(arr) <= 1:
    return arr

  mid = len(arr)//2
  left = mergesort(arr[:mid])
  right = mergesort(arr[mid:])

  left_sorted = mergesort(left)
  right_sorted = mergesort(right)

  return merge(left_sorted, right_sorted)


def merge(left, right):
  sorted_array = []
  left_index, right_index = 0, 0

  while left_index < len(left) and right_index < len(right):
    if left[left_index] < right[right_index]:
      sorted_array.append(left[left_index])
      left_index += 1
    else:
      sorted_array.append(right[right_index])
      right_index += 1

  while left_index < len(left):
    sorted_array.append(left[left_index])
    left_index += 1

  while right_index < len(right):
    sorted_array.append(right[right_index])
    right_index += 1

  return sorted_array
```

```python
arr6 = [1,45,67,32,24,89]
print(mergesort(arr6))
```

**11. Quick Sort**

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]  # Choose the middle element as the pivot
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)

# Example usage
arr = [38, 27, 43, 3, 9, 82, 10]
sorted_arr = quick_sort(arr)
print(f"Original array: {arr}")
print(f"Sorted array: {sorted_arr}")
```

```python
arr7 = [1,45,67,32,24,89]
print(quick_sort(arr7))
```

STRING PROBLEMS

```python
def count_character(s):
  vowel , consonant, space = 0, 0, 0
  for char in s:
    if char.lower() in 'aeiou':
      vowel += 1
    elif char.isspace():
      space += 1
    else:
      consonant += 1
  return vowel, consonant, space
```

```python
s1 = "hello world"
print(count_character(s1))
```

Check if string is palindrome or not:

```python
def ispalindrome(s):
  start, end = 0, len(s)-1
  while start <= end:
    if s[start] != s[end]:
      return False
    start += 1
    end -= 1
  return True
```

```python
s1 = "malayalam"
print(ispalindrome(s1))
```

Remove all the vowels from a string:

```python
def remove_vowels(s):
  vowels = 'aeiouAEIOU'
  result = ""
  for char in s:
    if char not in vowels:
      result += char
  return result
```

```python
str = "Hello World"
print(remove_vowels(str))
```

Remove spaces from a string:

```python
def remove_spaces(s):
  result = ""
  for char in s:
    if char != ' ':
      result += char
  return result
```

```python
str = "Hello World"
print(remove_spaces(str))
```

```python
def removeSpace(s):
  return s.replace(' ', '')
```

```python
str = "Hello World"
print(removeSpace(str))
```

Remove characters from a string except alphabets:

```python
def remove_char(s):
  result = ""
  for char in s:
    if char.isalpha():
      result += char
  return result
```

```python
str = "Hello123World"
print(remove_char(str))
```

```python
def non_alphabet(s):
  return ''.join([char for char in s if char.isalpha()])
```

```python
s1 = "Hello123World"
print(non_alphabet(s1))
```

Reverse  a string:


```python
def reverse_string(s):
  result = ""
  # reverse = s[::-1]
  for i in range(len(s)-1, -1, -1):
    result += s[i]
  return result
```

```python
s1 = "Hello World"
print(reverse_string(s1))
```

Remove brackets from an algebric expression:

```python
def remove_brackets(s):
  result = ""
  for c in s:
    if c != '(' and c != ')':
      result += c
  return result
```

```python
s1 = "(a+b)*c"
print(remove_brackets(s1))
```

Calculate sum of the numbers in a string:

```python
def sum_numbers(s):
  result = 0
  num = ""
  for char in s:
    if char.isdigit():
      num += char
    elif num:
      result += int(num)
      num = ""
  if num:
    result += int(num)
  return result
```

```python
s2 = "123abc456def"
print(sum_numbers(s2))
```
